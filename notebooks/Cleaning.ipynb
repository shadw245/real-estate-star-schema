```python
df=df.drop(columns=['prev_sold_date'])
```

```python

df = df.dropna(subset=[
    'price',
    'city',
    'state',
    'zip_code',
    'street',
    'brokered_by'
])
```

```python
df['bed'] = df['bed'].fillna(df['bed'].median())
```

```python
df['bath'] = df['bath'].fillna(df['bath'].median())
```

```python
print(df['house_size'].describe())
print(df['acre_lot'].describe())
```

```python
df['house_size'] = df['house_size'].fillna(df['house_size'].median())
```

```python
df['acre_lot'] = df['acre_lot'].fillna(df['acre_lot'].median())
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
#check and remove duplicates
duplicates = df.duplicated().sum()
print(f"Duplicate rows: {duplicates}")
```

```python
df = df.drop_duplicates()
```

```python
df['price'] = df['price'].astype(int)
df['house_size'] = df['house_size'].astype(int)
df['bed'] = df['bed'].astype(int)
df['bath'] = df['bath'].astype(int)
df['zip_code'] = df['zip_code'].astype(int)
df['brokered_by']=df['brokered_by'].astype(int)

print("Data types after conversion:")
display(df[['price', 'house_size', 'bed', 'bath', 'zip_code','brokered_by']].dtypes)
```

```python

print("Price value counts:")
display(df['price'].value_counts())
print("\nBed value counts:")
display(df['bed'].value_counts())
print("\nBath value counts:")
display(df['bath'].value_counts())
print("\nAcre_lot value counts:")
display(df['acre_lot'].value_counts())
print("\nHouse_size value counts:")
display(df['house_size'].value_counts())
```

```python
for col in ['price', 'bed', 'bath', 'acre_lot', 'house_size']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

print(f"DataFrame shape after outlier removal: {df.shape}")
```

```python
print("Price value counts:")
display(df['price'].value_counts())
print("\nBed value counts:")
display(df['bed'].value_counts())
print("\nBath value counts:")
display(df['bath'].value_counts())
print("\nAcre_lot value counts:")
display(df['acre_lot'].value_counts())
print("\nHouse_size value counts:")
display(df['house_size'].value_counts())
```
